```python
client = AzureOpenAI(
    azure_endpoint= AZURE_OPENAI_ENDPOINT,
    api_key=AZURE_OPENAI_KEY,
    api_version=API_VERSION
)
```

```python
response = client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": "I am going to Paris, what should I see?",
        }
    ],
    max_tokens=4096,
    temperature=1.0,
    top_p=1.0,
    model=DEPLOYMENT
)
```

Pour information:
| Hyperparamètre  | Rôle                             | Valeur 1.0 signifie quoi ?                                |
| --------------- | -------------------------------- | --------------------------------------------------------- |
| **max_tokens**  | Limite la longueur de la réponse | La réponse peut être *longue*, mais limitée à 4096 tokens |
| **temperature** | contrôle la créativité           | niveau standard → réponses variées, naturelles            |
| **top_p**       | filtre les tokens possibles      | toutes les probabilités sont autorisées                   |


```python
print(response.choices[0].message.content)
```

```python
#Pour un flow de réponse en continu
response = client.chat.completions.create(
    stream=True,
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": "I am going to Paris, what should I see?",
        }
    ],
    max_tokens=4096,
    temperature=1.0,
    top_p=1.0,
    model=DEPLOYMENT,
)

for update in response:
    if update.choices:
        print(update.choices[0].delta.content or "", end="")

client.close()
```

Analyse d'un dataset de clients, liées au projet de Churn

```python
import kagglehub
from pathlib import Path
# Download latest version
path = kagglehub.dataset_download("rivalytics/saas-subscription-and-churn-analytics-dataset")
# chargement des données ratvenstack_accounts.csv qui se trouve dans le path téléchargé
df = pd.read_csv(Path(path) / "ravenstack_accounts.csv")
```

```python
response = client.chat.completions.create(
    stream = True,
    messages=[
        {"role": "system", "content": "Tu es un expert data science."},
        {"role": "user", "content": f"Analyse ce dataset :\n{df.head().to_string()}"}
    ],
    max_tokens=4096,
    temperature=1.0,
    top_p=1.0,
    model=DEPLOYMENT,
)

for update in response:
    if update.choices:
        print(update.choices[0].delta.content or "", end="")

#client.close()
```
